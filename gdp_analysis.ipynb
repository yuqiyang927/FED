# GDP Analysis

This notebook explores GDP trends in the United Kingdom, United States,Brazil, Japan, China, Germany, and Switzerland from 2000 to 2022.

The horizontal axis shows the year, and the vertical axis shows GDP
in trillions of constant US dollars. The legend identifies each country.

GDP totals describe the size of an economy; they do not directly
measure average income or living standards.

## Data Lodaing and Cleaning

The data comes from the World Bank indicator `NY.GDP.MKTP.KD`, which measures GDP in constant US dollars. The price base year should be checked in the dataset metadata.

The analysis uses functions imported from separate Python modules to:

1. Load the original World Bank CSV file.
2. Select the seven countries and years 2000–2022.
3. Reshape the data so each row represents one country in one year for graphic analysis.

```python
# Below functions are used to load and clean GDP data from the Federal Reserve Economic Data (FRED) database.
from fed.data import load_gdp_data, clean_gdp_data

raw_data = load_gdp_data()
df = clean_gdp_data(raw_data)
```

## Plotting Annual GDP by Year for Seven Countries (2000–2022)

The chart compares annual GDP in the United Kingdom, United States,
Brazil, Japan, China, Germany, and Switzerland.

- **X-axis:** Year, from 2000 to 2022.
- **Y-axis:** GDP in trillions of constant US dollars.
- **Colours:** Each country is represented by a different colour.

```python
# Below function is used to plot the cleaned GDP data.
from fed.plot_utils import plot_gdp_data

fig, ax = plot_gdp_data(df)
```
